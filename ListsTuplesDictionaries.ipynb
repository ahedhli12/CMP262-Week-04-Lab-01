## Part 3 – Dictionaries

A **dictionary** stores data using **key-value pairs**.

### Example
```python
person = {
    "name": "Alex",
    "age": 20
}

print(person)
print(person["name"])
```

A key works like a label. Here, `"name"` is the key and `"Alex"` is the value.

````python
person = {
    "name": "Alex",
    "age": 20
}

print(person)
print(person["name"])
````
