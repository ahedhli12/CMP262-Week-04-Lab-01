## Part 3 – Dictionaries

### Reminder
Use the pattern shown earlier in class. Do not copy an answer from the README.

```python

    "age": 20
}

```

A key works like a label. Here, `"name"` is the key and `"Alex"` is the value.

````python
person = {
    "name": "Alex",
    "age": 20
}

print(person)
print(person["name"])
````
